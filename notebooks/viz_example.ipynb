```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the table
judges = ['Paired' ,'Single', 'JUSSA']
base_scores = [1, 2, 2]
provoked_scores = [5, 3, 7]

# Set up the figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# Set the width of bars and positions
bar_width = 0.35
x_pos = np.arange(len(judges))

# Create the bars
bars1 = ax.bar(x_pos - bar_width/2, base_scores, bar_width, 
               label='Base', color='#2E86AB', alpha=0.8)
bars2 = ax.bar(x_pos + bar_width/2, provoked_scores, bar_width, 
               label='Provoked', color='#A23B72', alpha=0.8)

# Customize the plot
ax.set_xlabel('Judge Type', fontsize=12, fontweight='bold')
ax.set_ylabel('Score', fontsize=12, fontweight='bold')
ax.set_title('Score Comparison: Base vs Provoked Responses by Judge Type', 
             fontsize=14, fontweight='bold', pad=20)
ax.set_xticks(x_pos)
ax.set_xticklabels(judges)

# Add value labels on bars
def add_value_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom',
                    fontweight='bold')

add_value_labels(bars1)
add_value_labels(bars2)

# Add legend
ax.legend(loc='upper left', frameon=True, shadow=True)

# Add grid for better readability
ax.grid(axis='y', alpha=0.3, linestyle='--')
ax.set_axisbelow(True)

# Set y-axis to start from 0 and add some padding
ax.set_ylim(0, max(max(base_scores), max(provoked_scores)) * 1.2)

# Improve layout
plt.tight_layout()

# Add a subtle background color
ax.set_facecolor('#f8f9fa')

# Calculate and display differences
print("Score Differences (Provoked - Base):")
for i, judge in enumerate(judges):
    diff = provoked_scores[i] - base_scores[i]
    print(f"{judge}: {diff:+d}")

# Display the plot
plt.show()

# Optional: Save the figure
# plt.savefig('judge_scores_comparison.png', dpi=300, bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import numpy as np

def scale_fonts(scale_factor=1.0):
    """
    Function to easily scale font sizes throughout the plot
    scale_factor: multiplier for all font sizes (1.0 = default, 1.5 = 50% larger, etc.)
    """
    return {
        'xlabel_size': int(12 * scale_factor),
        'ylabel_size': int(12 * scale_factor),
        'title_size': int(14 * scale_factor),
        'tick_size': int(10 * scale_factor),
        'legend_size': int(10 * scale_factor),
        'annotation_size': int(9 * scale_factor)
    }

def create_judge_comparison_plot(judges, base_scores, provoked_scores, 
                               title="Manipulation scores \n by Judge Type",
                               figsize=(10, 6), scale_factor=1.0, 
                               base_color='#2E86AB', provoked_color='#A23B72',
                               save_path=None, show_plot=True):
    """
    Create a bar chart comparing base vs provoked scores for different judge types.
    
    Parameters:
    judges (list): List of judge type names
    base_scores (list): List of base scores corresponding to each judge
    provoked_scores (list): List of provoked scores corresponding to each judge
    title (str): Title for the plot
    figsize (tuple): Figure size (width, height)
    scale_factor (float): Font scaling factor
    base_color (str): Color for base score bars
    provoked_color (str): Color for provoked score bars
    save_path (str): Path to save the figure (optional)
    show_plot (bool): Whether to display the plot
    
    Returns:
    tuple: (fig, ax) matplotlib figure and axis objects
    """
    
    # Set up the figure and axis
    fig, ax = plt.subplots(figsize=figsize)

    # Get font sizes
    fonts = scale_fonts(scale_factor=scale_factor)

    # Set the width of bars and positions
    bar_width = 0.2
    x_pos = np.arange(len(judges))*0.6

    x_pos_2 = x_pos  + 0.01

    # Create the bars
    bars1 = ax.bar(x_pos - bar_width/2, base_scores, bar_width, 
                   label='Base', color=base_color, alpha=0.8)
    bars2 = ax.bar(x_pos_2 + bar_width/2, provoked_scores, bar_width, 
                   label='Provoked', color=provoked_color, alpha=0.8)

    # Customize the plot
    ax.set_xlabel('Judge Type', fontsize=fonts['xlabel_size'], fontweight='bold')
    ax.set_ylabel('Score', fontsize=fonts['ylabel_size'], fontweight='bold')
    ax.set_title(title, fontsize=fonts['title_size'], fontweight='bold', pad=10)
    ax.set_xticks(x_pos)
    ax.set_xticklabels(judges)

    # Set tick label sizes
    ax.tick_params(axis='both', which='major', labelsize=fonts['tick_size'])

    # Add value labels on bars
    def add_value_labels(bars):
        for bar in bars:
            height = bar.get_height()
            ax.annotate(f'{height}',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom',
                        fontweight='bold', fontsize=fonts['annotation_size'])

    # add_value_labels(bars1)
    # add_value_labels(bars2)

    # Add legend
    ax.legend(loc='upper left', frameon=True, shadow=True, fontsize=fonts['legend_size'])

    # Remove grid lines
    ax.set_axisbelow(True)

    # Set y-axis to start from 0 and add some padding
    ax.set_ylim(0, max(max(base_scores), max(provoked_scores)) * 1.1)

    # Improve layout
    plt.tight_layout()

    # Add a subtle background color
    ax.set_facecolor('#f8f9fa')

    # Calculate and display differences
    print("Score Differences (Provoked - Base):")
    for i, judge in enumerate(judges):
        diff = provoked_scores[i] - base_scores[i]
        print(f"{judge}: {diff:+d}")

    # Save the figure if path is provided
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"Figure saved to: {save_path}")

    # Display the plot
    if show_plot:
        plt.show()
    
    return fig, ax

# Example usage with the original data
judges = ['Single', 'Paired', 'JUSSA']
base_scores = [2, 1, 2]
provoked_scores = [3, 5, 7]

base_color = "#366eb7"
provoked_color = "#d6281f"

# Create the plot using the function
fig, ax = create_judge_comparison_plot(judges, base_scores, provoked_scores, scale_factor=2.1, figsize=(6, 5), base_color=base_color, provoked_color=provoked_color)


# save fig as pdf
fig.savefig("./figs/example_judge_scores_bars.svg", bbox_inches='tight')
```
